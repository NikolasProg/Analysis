```python
import pandas as pd
from pandasql import sqldf

# Читаем данные
df = pd.read_csv('data.tsv', sep='\t', header=None)
df.columns = ['query', 'timestamp', 'platform']
df.head()
```

```python
# Добавляем datetime
df['datetime'] = pd.to_datetime(df['timestamp'], unit='s')
df['date'] = df['datetime'].dt.date
df['hour'] = df['datetime'].dt.hour
df.head()
```

```python
pysqldf = lambda q: sqldf(q, globals())

# Диапазон дат
range = pysqldf("""
    SELECT
        MIN(date) as Начальная_дата,
        MAX(date) as Конечная_дата
    FROM df
""")
print(range)
```

```python
# Запросы с «ютуб» по платформам
YouTube = pysqldf("""
    SELECT
        platform as Платформа,
        COUNT(*) as Количество
    FROM df
    WHERE query LIKE '%ютуб%'
    GROUP BY platform
""")
print(YouTube)
```

Компьютеры, инструмент для продуктивности: пользователи ищут учебные материалы, справочники, рабочие инструменты. Мобильные устройства, для жизни: здесь люди поздравляют близких, развлекаются, следят за новостями и погодой.

```python
# Топ-10 запросов по платформам
top10 = pysqldf("""
    SELECT Платформа, Запрос, Количество
    FROM (
        SELECT
            platform as Платформа,
            query as Запрос,
            COUNT(*) as Количество,
            ROW_NUMBER() OVER (PARTITION BY platform ORDER BY COUNT(*) DESC) as Номер
        FROM df
        GROUP BY platform, query
    )
    WHERE Номер <= 10
    ORDER BY Платформа, Количество DESC
""")
print(top10)
```

```python
# Трафик по часам
hourly = pysqldf("""
    SELECT
        platform as Платформа,
        hour as Час,
        COUNT(*) as Количество
    FROM df
    GROUP BY platform, hour
    ORDER BY platform, hour
""")
print(hourly)
```

Из данных мы видим что пользователи телефонов активно используют поисковик утром и вечером, можно предполоить что люди утром просыпаются и смотрят погоду и ищут прочую информацию, а так же вечером приходят с работы и снова залазят в телефон. А пользователи с компьютерами ведут активность в дневное время, скорее всего находясь на работе.

```python
# График трафика по часам
hourly_pivot = hourly.pivot(index='Час', columns='Платформа', values='Количество').fillna(0)
hourly_pivot.plot(kind='line', title='Запросы по часам')
```

```python
# Тематики запросов
topic_counts = pysqldf("""
    SELECT
        platform as Платформа,
        CASE
            WHEN query LIKE '%кот%' OR query LIKE '%собака%' OR query LIKE '%алабай%' THEN 'животные'
            WHEN query LIKE '%мем%' OR query LIKE '%прикол%' THEN 'мемы'
            WHEN query LIKE '%обои%' OR query LIKE '%фон%' THEN 'обои'
            WHEN query LIKE '%еда%' OR query LIKE '%пицца%' OR query LIKE '%кофе%' THEN 'еда'
            WHEN query LIKE '%актёр%' OR query LIKE '%певец%' OR query LIKE '%знаменитости%' THEN 'звезды'
            ELSE 'другое'
        END as Тема,
        COUNT(*) as Количество
    FROM df
    GROUP BY platform, Тема
""")
print(topic_counts)
```

```python
# Доли
topic_share = topic_counts.pivot(index='Тема', columns='Платформа', values='Количество').fillna(0)
topic_share['Доля_компьютер'] = topic_share['desktop'] / topic_share['desktop'].sum()
topic_share['Доля_телефон'] = topic_share['touch'] / topic_share['touch'].sum()
topic_share['Разница'] = topic_share['Доля_компьютер'] - topic_share['Доля_телефон']
topic_share.sort_values('Разница')
```
